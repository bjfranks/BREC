```python
import pickle
import torch
import io
import os
```

```python
class CPU_Unpickler(pickle.Unpickler):
    def find_class(self, module, name):
        if module == 'torch.storage' and name == '_load_from_bytes':
            return lambda b: torch.load(io.BytesIO(b), map_location='cpu')
        else: return super().find_class(module, name)
```

```python
directory = os.fsencode("DropGNN/results/")
data_dict = dict()
for file in os.listdir(directory):
    filename = os.fsdecode(file)
    if filename.endswith(".pkl"):
        with open(os.path.join(os.fsdecode(directory), filename), 'rb') as f:
            while True:
                try:
                    data = CPU_Unpickler(f).load()
                except EOFError:
                    break
                if data[-1].augmentation == "PSE":
                    data_dict[(data[-1].pse, data[1][0], data[-1].loss, data[-1].loss_parameter)] = data[:-1] #(data[-1].EPOCH, data[-1].LEARNING_RATE, data[-1].WEIGHT_DECAY, data[-1].hidden_units, data[-1].num_layers, data[-1].added_dimensions)
                if data[-1].augmentation == "random":
                    data_dict[(data[-1].random, data[1][0], data[-1].loss, data[-1].loss_parameter)] = data[:-1]
                if data[-1].augmentation == "rewiring":
                    data_dict[(data[-1].rewire, data[1][0], data[-1].loss, data[-1].loss_parameter)] = data[:-1]
                if data[-1].augmentation == "none":
                    data_dict[("none", data[1][0], data[-1].loss, data[-1].loss_parameter)] = data[:-1]
print(data_dict.keys())
```

```python
keys = data_dict.keys()
counts = dict()
for key in keys:
    value = data_dict[key]
    counts[key] = sum([1 for x in value if x[2] and x[4]])
for key, value in counts.items():
    print(value, key)
#print(hp_search)
```

```python
for method in ["none", "RWSE", "SPDPE", "RDPE", "CycleSE", "LapPE", "RLapPE", "ElstaticPE", "RElstaticPE", "HKdiagSE", "RHKdiagSE", "gaussian", "RNI", 
               "binary", "IRNI", "DE", "AE", "CGP", "EGP"]:
    for loss in ["CosineEmbeddingLoss", "nt_bxent_loss"]:
        for lp in [1, 0.5]:
            for part in ["Basic", "Regular", "Extension", "CFI", "4-Vertex_Condition", "Distance_Regular", "CCoHG", "3r2r", "pep"]:
                key = (method, part, loss, lp)
                if key in counts:
                    print(counts[key], key)
```

```python
for loss in ["CosineEmbeddingLoss", "nt_bxent_loss"]:
    for lp in [1, 0.5]:
        print(loss, lp) #, "ElstaticPE", "HKdiagSE"
        for method in ["none", "RWSE", "SPDPE", "RDPE", "CycleSE", "LapPE", "RLapPE", "RElstaticPE", "RHKdiagSE", "gaussian", 
                       "RNI", "binary", "IRNI", "DE", "AE", "EGP", "CGP"]:
            print(method, "& ", end="")
            count1 = 0
            count2 = 0
            for i, (part, count) in enumerate([("Regular", 100), ("4-Vertex_Condition", 20), ("Distance_Regular", 20)]):
                if i>0:
                    print(" & ", end="")
                count2+=count
                key = (method, part, loss, lp)
                if key in counts:
                    count1+=counts[key]
                    print(counts[key], "&", str((100*counts[key])//count)+"\\%", end="")
                else:
                    print("-", "&", "-", end="")
            print(" &", count1, "&", str((100*count1)//count2)+"\\%\\\\")
        print()
```
